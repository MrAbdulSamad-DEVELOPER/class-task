```python
def analyze_transactions(transactions):
    seen = set()
    totals = {}
    cats = {}
    customers = []
    for tx in transactions:
        for key in ("id", "customer", "category", "amount", "status"):
            if key not in tx:
                raise ValueError("Missing key: " + key)
        if tx["id"] in seen:
            raise ValueError("Duplicate transaction ID: " + str(tx["id"]))
        seen.add(tx["id"])
        c = tx["customer"]
        if not isinstance(c, str) or not c.strip():
            raise ValueError("Customer name must be a non-empty string")
        if c not in totals:
            totals[c] = 0
            cats[c] = {}
            customers.append(c)
        a = tx["amount"]
        if isinstance(a, bool) or not isinstance(a, (int, float)) or a < 0:
            raise ValueError("Amount must be a non-negative number")
        if tx["status"] == "success":
            totals[c] += a
            cats[c][tx["category"]] = cats[c].get(tx["category"], 0) + a
    customers.sort(key=lambda c: (-totals[c], c))
    report = []
    for rank, c in enumerate(customers, 1):
        top = None
        best = -1
        for cat in sorted(cats[c]):
            if cats[c][cat] > best:
                best = cats[c][cat]
                top = cat
        report.append({"customer": c, "total_spending": totals[c], "top_category": top, "rank": rank})
    return report


if __name__ == "__main__":
    data = [
        {"id": 1, "customer": "Ali", "category": "Food", "amount": 500, "status": "success"},
        {"id": 2, "customer": "Ali", "category": "Travel", "amount": 200, "status": "failed"},
        {"id": 3, "customer": "Sara", "category": "Food", "amount": 700, "status": "success"},
    ]
    for row in analyze_transactions(data):
        print(row)
```

```python
def longest_unique_window(text):
    last = {}
    start = 0
    best_start = 0
    best_len = 0
    for i, ch in enumerate(text):
        if ch in last and last[ch] >= start:
            start = last[ch] + 1
        last[ch] = i
        if i - start + 1 > best_len:
            best_start, best_len = start, i - start + 1
    return text[best_start:best_start + best_len], best_start, best_len


if __name__ == "__main__":
    print(longest_unique_window("abcabcbb"))
    print(longest_unique_window(""))
    print(longest_unique_window("bbbbb"))
    print(longest_unique_window("pwwkew"))
```

```python
def find_paths(data, target):
    paths = []

    def same(a, b):
        if isinstance(a, bool) or isinstance(b, bool):
            return a is b
        if isinstance(a, (int, float)) and isinstance(b, (int, float)):
            return a == b
        return type(a) == type(b) and a == b

    def walk(value, path):
        if same(value, target):
            paths.append(path or "$")
        if isinstance(value, dict):
            for key, child in value.items():
                walk(child, path + "." + str(key) if path else str(key))
        elif isinstance(value, list):
            for i, child in enumerate(value):
                walk(child, path + "[" + str(i) + "]")

    walk(data, "")
    return paths


if __name__ == "__main__":
    print(find_paths({"a": [{"x": 5}, 5], "b": {"y": 5}}, 5))
    print(find_paths({"x": True}, 1))
```

```python
def execution_order(tasks, dependencies):
    prereq = {t: set() for t in tasks}
    for before, after in dependencies:
        if before not in prereq or after not in prereq:
            raise ValueError("Unknown task: " + str((before, after)))
        prereq[after].add(before)
    done = []
    remaining = list(tasks)
    while remaining:
        ready = sorted(t for t in remaining if prereq[t] <= set(done))
        if not ready:
            raise ValueError("Dependency cycle detected")
        task = ready[0]
        done.append(task)
        remaining.remove(task)
    return done


if __name__ == "__main__":
    print(execution_order(["build", "test", "deploy"], [("build", "test"), ("test", "deploy")]))
    print(execution_order(["a", "b", "c"], [("a", "c")]))
    try:
        execution_order(["a", "b"], [("a", "b"), ("b", "a")])
    except ValueError as e:
        print("Error:", e)
```

```python
class Parser:
    def __init__(self, text):
        self.tokens = []
        i = 0
        while i < len(text):
            ch = text[i]
            if ch == " ":
                i += 1
            elif ch.isdigit():
                j = i
                while j < len(text) and text[j].isdigit():
                    j += 1
                self.tokens.append(text[i:j])
                i = j
            elif ch in "+-*/()":
                self.tokens.append(ch)
                i += 1
            else:
                raise ValueError("Invalid character: " + ch)
        self.pos = 0

    def peek(self):
        return self.tokens[self.pos] if self.pos < len(self.tokens) else None

    def take(self):
        token = self.peek()
        self.pos += 1
        return token

    def parse(self):
        value = self.expr()
        if self.peek() is not None:
            raise ValueError("Unexpected token: " + str(self.peek()))
        return value

    def expr(self):
        value = self.term()
        while self.peek() in ("+", "-"):
            op = self.take()
            right = self.term()
            value = value + right if op == "+" else value - right
        return value

    def term(self):
        value = self.unary()
        while self.peek() in ("*", "/"):
            op = self.take()
            right = self.unary()
            if op == "/" and right == 0:
                raise ZeroDivisionError("division by zero")
            value = value * right if op == "*" else value / right
        return value

    def unary(self):
        if self.peek() == "+":
            self.take()
            return self.unary()
        if self.peek() == "-":
            self.take()
            return -self.unary()
        return self.atom()

    def atom(self):
        token = self.take()
        if token is None:
            raise ValueError("Expected a number or parenthesis")
        if token == "(":
            value = self.expr()
            if self.take() != ")":
                raise ValueError("Missing closing parenthesis")
            return value
        if token.isdigit():
            return int(token)
        raise ValueError("Unexpected token: " + token)


def evaluate_expression(text):
    return Parser(text).parse()


if __name__ == "__main__":
    print(evaluate_expression("-(3 + 2) * 4 + 18 / 3"))
    print(evaluate_expression("2 + 3 * 4"))
    print(evaluate_expression("(2 + 3) * 4"))
    try:
        evaluate_expression("1/0")
    except ZeroDivisionError as e:
        print("Error:", e)
```

```python
import uuid


class Account:
    def __init__(self, account_id, balance=0):
        if balance < 0:
            raise ValueError("Balance must be 0 or more")
        self.account_id = account_id
        self._balance = balance
        self._transactions = []

    @property
    def balance(self):
        return self._balance

    def _record(self, action, amount):
        self._transactions.append({"transaction_id": uuid.uuid4().hex, "action": action,
                                   "amount": amount, "resulting_balance": self._balance})

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        self._balance += amount
        self._record("deposit", amount)

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        if amount > self._balance:
            raise ValueError("Insufficient funds")
        self._balance -= amount
        self._record("withdrawal", amount)

    def statement(self):
        return list(self._transactions)


class Bank:
    def __init__(self):
        self.accounts = {}

    def add_account(self, account):
        if account.account_id in self.accounts:
            raise ValueError("Duplicate account ID")
        self.accounts[account.account_id] = account

    def get(self, account_id):
        if account_id not in self.accounts:
            raise KeyError("Account not found: " + str(account_id))
        return self.accounts[account_id]

    def transfer(self, source_id, dest_id, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        source = self.get(source_id)
        dest = self.get(dest_id)
        if source is dest:
            raise ValueError("Cannot transfer to the same account")
        if source.balance < amount:
            raise ValueError("Insufficient funds")
        source._balance -= amount
        dest._balance += amount
        source._record("transfer_out:" + dest_id, amount)
        dest._record("transfer_in:" + source_id, amount)


if __name__ == "__main__":
    a, b = Account("A", 1000), Account("B", 200)
    bank = Bank()
    bank.add_account(a)
    bank.add_account(b)
    bank.transfer("A", "B", 350)
    print("A:", a.balance, "B:", b.balance)
    print(a.statement())
    print(b.statement())
```

```python
from abc import ABC, abstractmethod


class Employee(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def calculate_pay(self):
        pass


class SalariedEmployee(Employee):
    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary

    def calculate_pay(self):
        return self.salary


class HourlyEmployee(Employee):
    def __init__(self, name, rate, hours):
        super().__init__(name)
        self.rate = rate
        self.hours = hours

    def calculate_pay(self):
        return min(self.hours, 40) * self.rate + max(0, self.hours - 40) * self.rate * 1.5


class CommissionEmployee(Employee):
    def __init__(self, name, base, sales, rate):
        super().__init__(name)
        self.base = base
        self.sales = sales
        self.rate = rate

    def calculate_pay(self):
        return self.base + self.sales * self.rate


def payroll_summary(employees):
    return [{"name": e.name, "pay": e.calculate_pay()} for e in employees]


if __name__ == "__main__":
    staff = [SalariedEmployee("Ali", 50000), HourlyEmployee("Sara", 20, 45),
             CommissionEmployee("Ahmed", 500, 4000, 0.10)]
    for row in payroll_summary(staff):
        print(row)
```

```python
import time
from functools import wraps


def ttl_cache(seconds):
    def decorator(func):
        cache = {}
        stats = {"hits": 0, "misses": 0}

        @wraps(func)
        def wrapper(*args, **kwargs):
            key = (args, tuple(sorted(kwargs.items())))
            now = time.monotonic()
            if key in cache and now < cache[key][0]:
                stats["hits"] += 1
                return cache[key][1]
            stats["misses"] += 1
            value = func(*args, **kwargs)
            cache[key] = (now + seconds, value)
            return value

        wrapper.cache_info = lambda: dict(stats)
        wrapper.cache_clear = lambda: (cache.clear(), stats.update(hits=0, misses=0))
        return wrapper
    return decorator


@ttl_cache(10)
def add(x, y):
    return x + y


if __name__ == "__main__":
    print(add(x=2, y=3))
    print(add(y=3, x=2))
    print(add(2, 3))
    print(add(2, 3))
    print(add.cache_info())
```

```python
import os
from datetime import datetime


def read_errors(path, malformed=None):
    if malformed is None:
        malformed = {"count": 0}
    with open(path, "r", encoding="utf-8") as f:
        for line_number, line in enumerate(f, 1):
            parts = line.rstrip("\n").split(" | ")
            if len(parts) != 3:
                malformed["count"] += 1
                continue
            timestamp, level, message = parts
            try:
                datetime.strptime(timestamp, "%Y-%m-%d %H:%M:%S")
            except ValueError:
                malformed["count"] += 1
                continue
            if level.strip().upper() == "ERROR":
                yield line_number, timestamp, message


def summarize_errors(path):
    counts = {}
    malformed = {"count": 0}
    for _, timestamp, _ in read_errors(path, malformed):
        date = timestamp[:10]
        counts[date] = counts.get(date, 0) + 1
    return dict(sorted(counts.items())), malformed["count"]


if __name__ == "__main__":
    # Automatically create a dummy app.log file to prevent FileNotFoundError
    if not os.path.exists("app.log"):
        with open("app.log", "w", encoding="utf-8") as dummy_log:
            dummy_log.write("2026-11-07 10:15:30 | ERROR | Connection timed out\n")
            dummy_log.write("2026-11-07 10:16:45 | INFO | Server started\n")
            dummy_log.write("Malformed log line without pipes\n")
            dummy_log.write("2026-11-08 12:00:00 | ERROR | Database connection failed\n")

    bad = {"count": 0}
    for line_no, ts, msg in read_errors("app.log", bad):
        print(line_no, ts, msg)
    print("Malformed:", bad["count"])
    print(summarize_errors("app.log"))
```

```python
import csv
import os


def reconcile_inventory(inventory_path, movements_path, updated_path, rejected_path):
    stock = {}
    names = {}
    with open(inventory_path, newline="", encoding="utf-8") as f:
        for row in csv.DictReader(f):
            sku = row["sku"].strip()
            qty = int(row["quantity"])
            if not sku or qty < 0 or sku in stock:
                raise ValueError("Invalid or duplicate SKU: " + sku)
            stock[sku] = qty
            names[sku] = row["name"]

    rejected = []
    with open(movements_path, newline="", encoding="utf-8") as f:
        for row_num, row in enumerate(csv.DictReader(f), 2):
            sku = row.get("sku", "").strip()
            kind = row.get("type", "").strip().upper()
            try:
                units = int(row.get("units", ""))
            except ValueError:
                units = 0
            reason = None
            if sku not in stock:
                reason = "Unknown SKU"
            elif units <= 0:
                reason = "Units must be positive"
            elif kind not in ("IN", "OUT"):
                reason = "Invalid movement type"
            elif kind == "OUT" and stock[sku] - units < 0:
                reason = "Insufficient stock"
            if reason:
                bad = dict(row)
                bad["row_number"] = row_num
                bad["reason"] = reason
                rejected.append(bad)
                continue
            stock[sku] += units if kind == "IN" else -units

    with open(updated_path, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=["sku", "name", "quantity"])
        writer.writeheader()
        for sku in stock:
            writer.writerow({"sku": sku, "name": names[sku], "quantity": stock[sku]})

    fields = list(rejected[0].keys()) if rejected else ["row_number", "sku", "type", "units", "reason"]
    with open(rejected_path, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=fields)
        writer.writeheader()
        writer.writerows(rejected)
    return stock, rejected


if __name__ == "__main__":
    # Generate dummy input files if they don't exist
    if not os.path.exists("inventory.csv"):
        with open("inventory.csv", "w", newline="", encoding="utf-8") as f:
            writer = csv.DictWriter(f, fieldnames=["sku", "name", "quantity"])
            writer.writeheader()
            writer.writerows([
                {"sku": "SKU001", "name": "Item A", "quantity": 10},
                {"sku": "SKU002", "name": "Item B", "quantity": 5}
            ])

    if not os.path.exists("movements.csv"):
        with open("movements.csv", "w", newline="", encoding="utf-8") as f:
            writer = csv.DictWriter(f, fieldnames=["sku", "type", "units"])
            writer.writeheader()
            writer.writerows([
                {"sku": "SKU001", "type": "IN", "units": 5},
                {"sku": "SKU002", "type": "OUT", "units": 2},
                {"sku": "SKU001", "type": "OUT", "units": 20}  # Will be rejected (insufficient stock)
            ])

    stock, rejected = reconcile_inventory("inventory.csv", "movements.csv",
                                          "updated_inventory.csv", "rejected_movements.csv")
    print("Updated Stock:", stock)
    print("Rejected Movements:", rejected)
```

```python
import csv
import os


def build_gradebook(students_path, scores_path, report_path):
    students = {}
    with open(students_path, newline="", encoding="utf-8") as f:
        for row in csv.DictReader(f):
            sid = row["student_id"].strip()
            if not sid or sid in students:
                raise ValueError("Invalid or duplicate student ID: " + sid)
            students[sid] = row["name"]

    scores = {}
    seen = set()
    with open(scores_path, newline="", encoding="utf-8") as f:
        for row in csv.DictReader(f):
            sid = row["student_id"].strip()
            subject = row["subject"].strip()
            if sid not in students:
                raise ValueError("Unknown student ID: " + sid)
            if (sid, subject) in seen:
                raise ValueError("Duplicate student-subject: " + str((sid, subject)))
            seen.add((sid, subject))
            mark = float(row["marks"])
            if not 0 <= mark <= 100:
                raise ValueError("Marks must be between 0 and 100")
            scores.setdefault(sid, []).append(mark)

    def grade(avg):
        if avg >= 85:
            return "A"
        if avg >= 70:
            return "B"
        if avg >= 55:
            return "C"
        if avg >= 40:
            return "D"
        return "F"

    rows = []
    for sid, name in students.items():
        marks = scores.get(sid, [])
        avg = sum(marks) / len(marks) if marks else 0.0
        rows.append({"student_id": sid, "name": name, "average": round(avg, 2), "grade": grade(avg)})
    rows.sort(key=lambda r: (-r["average"], r["student_id"]))
    for rank, row in enumerate(rows, 1):
        row["rank"] = rank

    with open(report_path, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=["student_id", "name", "average", "grade", "rank"])
        writer.writeheader()
        writer.writerows(rows)
    return rows


if __name__ == "__main__":
    # Automatically create the required CSV files if they do not exist
    if not os.path.exists("students.csv"):
        with open("students.csv", "w", newline="", encoding="utf-8") as f:
            writer = csv.DictWriter(f, fieldnames=["student_id", "name"])
            writer.writeheader()
            writer.writerows([
                {"student_id": "101", "name": "Ayesha"},
                {"student_id": "102", "name": "Bilal"}
            ])

    if not os.path.exists("scores.csv"):
        with open("scores.csv", "w", newline="", encoding="utf-8") as f:
            writer = csv.DictWriter(f, fieldnames=["student_id", "subject", "marks"])
            writer.writeheader()
            writer.writerows([
                {"student_id": "101", "subject": "Math", "marks": "90"},
                {"student_id": "101", "subject": "English", "marks": "80"},
                {"student_id": "102", "subject": "Math", "marks": "65"},
                {"student_id": "102", "subject": "English", "marks": "75"}
            ])

    for row in build_gradebook("students.csv", "scores.csv", "report.csv"):
        print(row)
```

```python
class ReversibleRange:
    def __init__(self, start, stop, step=1):
        if step == 0:
            raise ValueError("step must not be zero")
        self.start, self.stop, self.step = start, stop, step

    def __iter__(self):
        current = self.start
        if self.step > 0:
            while current < self.stop:
                yield current
                current += self.step
        else:
            while current > self.stop:
                yield current
                current += self.step

    def __len__(self):
        if self.step > 0:
            return max(0, (self.stop - self.start + self.step - 1) // self.step)
        return max(0, (self.start - self.stop - self.step - 1) // (-self.step))

    def __reversed__(self):
        for i in range(len(self) - 1, -1, -1):
            yield self.start + i * self.step

    def __contains__(self, value):
        if isinstance(value, bool) or not isinstance(value, int):
            return False
        if self.step > 0 and not self.start <= value < self.stop:
            return False
        if self.step < 0 and not self.stop < value <= self.start:
            return False
        return (value - self.start) % self.step == 0


if __name__ == "__main__":
    r = ReversibleRange(2, 11, 3)
    print(list(r))
    print(list(reversed(r)))
    print(5 in r)
    print(len(r))
```

```python
from datetime import datetime


class BookingScheduler:
    def __init__(self):
        self.bookings = []
        self.ids = set()

    def add_booking(self, room, start, end, booking_id):
        if end <= start:
            return {"accepted": False, "reason": "End must be after start"}
        if booking_id in self.ids:
            return {"accepted": False, "reason": "Duplicate booking ID"}
        for b in self.bookings:
            if b["room"] == room and start < b["end"] and b["start"] < end:
                return {"accepted": False, "reason": "Booking overlaps an existing booking",
                        "conflict": b["booking_id"]}
        booking = {"room": room, "start": start, "end": end, "booking_id": booking_id}
        self.bookings.append(booking)
        self.ids.add(booking_id)
        self.bookings.sort(key=lambda b: (b["room"], b["start"], b["booking_id"]))
        return {"accepted": True, "booking": booking}


if __name__ == "__main__":
    s = BookingScheduler()
    d = datetime(2026, 11, 7)
    print(s.add_booking("A", d.replace(hour=9), d.replace(hour=10), "1")["accepted"])
    print(s.add_booking("A", d.replace(hour=10), d.replace(hour=11), "2")["accepted"])
    print(s.add_booking("A", d.replace(hour=9, minute=30), d.replace(hour=10, minute=30), "3"))
```

```python
import json
from pathlib import Path


def load_contacts(path):
    path = Path(path)
    if not path.exists():
        return {}
    try:
        data = json.loads(path.read_text(encoding="utf-8"))
    except (json.JSONDecodeError, UnicodeDecodeError):
        raise ValueError("Contact file is corrupt: " + str(path))
    if not isinstance(data, dict):
        raise ValueError("Contact data must be a JSON object")
    return data


def save_contacts(path, contacts):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    temp = path.with_suffix(path.suffix + ".tmp")
    temp.write_text(json.dumps(contacts, indent=2), encoding="utf-8")
    temp.replace(path)


def add_contact(contacts, contact_id, name, phone, email):
    contact_id = str(contact_id)
    if contact_id in contacts:
        raise ValueError("Contact ID already exists")
    if not name.strip() or not phone.strip() or not email.strip():
        raise ValueError("Name, phone, and email are required")
    contacts[contact_id] = {"id": contact_id, "name": name.strip(),
                            "phone": phone.strip(), "email": email.strip()}
    return contacts[contact_id]


def search_contacts(contacts, query):
    q = query.lower()
    return [c for c in contacts.values() if q in c["name"].lower()]


def update_contact(contacts, contact_id, **changes):
    contact_id = str(contact_id)
    if contact_id not in contacts:
        raise KeyError("Contact not found")
    for field in ("name", "phone", "email"):
        if field in changes:
            if not changes[field].strip():
                raise ValueError(field + " cannot be empty")
            contacts[contact_id][field] = changes[field].strip()
    return contacts[contact_id]


def delete_contact(contacts, contact_id):
    if str(contact_id) not in contacts:
        raise KeyError("Contact not found")
    return contacts.pop(str(contact_id))


def contact_manager_cli(path="contacts.json"):
    contacts = load_contacts(path)
    while True:
        print("\n1 Add  2 Search  3 Update  4 Delete  5 List  0 Exit")
        choice = input("Choose: ").strip()
        try:
            if choice == "1":
                add_contact(contacts, input("ID: "), input("Name: "),
                            input("Phone: "), input("Email: "))
                save_contacts(path, contacts)
            elif choice == "2":
                print(search_contacts(contacts, input("Name search: ")))
            elif choice == "3":
                update_contact(contacts, input("ID: "), name=input("New name: "))
                save_contacts(path, contacts)
            elif choice == "4":
                delete_contact(contacts, input("ID: "))
                save_contacts(path, contacts)
            elif choice == "5":
                print(list(contacts.values()))
            elif choice == "0":
                break
            else:
                print("Unknown option")
        except (ValueError, KeyError) as e:
            print("Error:", e)


if __name__ == "__main__":
    contacts = {}
    add_contact(contacts, "1", "Ayesha", "0300-1234567", "ayesha@example.com")
    add_contact(contacts, "2", "Bilal", "0300-7654321", "bilal@example.com")
    save_contacts("contacts.json", contacts)
    loaded = load_contacts("contacts.json")
    print(search_contacts(loaded, "aye"))
    print(list(loaded.values()))
```

```python
import json
import sys
import unittest
import uuid
from pathlib import Path


class Product:
    def __init__(self, product_id, name, price, stock):
        if not product_id or not name:
            raise ValueError("Product ID and name are required")
        if price < 0:
            raise ValueError("Price must be non-negative")
        if stock < 0:
            raise ValueError("Stock must be non-negative")
        self.product_id, self.name, self.price, self.stock = product_id, name, price, stock


class Inventory:
    def __init__(self, products):
        self.products = {}
        for p in products:
            if p.product_id in self.products:
                raise ValueError("Duplicate product ID")
            self.products[p.product_id] = p

    @classmethod
    def from_json(cls, path):
        data = json.loads(Path(path).read_text(encoding="utf-8"))
        return cls([Product(str(p["product_id"]), p["name"], p["price"], p["stock"]) for p in data])


class Cart:
    def __init__(self):
        self.items = {}

    def add(self, product_id, quantity):
        if quantity <= 0:
            raise ValueError("Quantity must be positive")
        self.items[product_id] = self.items.get(product_id, 0) + quantity

    def total(self, inventory, discount_percent=0):
        if not 0 <= discount_percent <= 100:
            raise ValueError("Discount must be between 0 and 100")
        subtotal = 0
        for pid, qty in self.items.items():
            if pid not in inventory.products:
                raise KeyError("Unknown product: " + pid)
            subtotal += inventory.products[pid].price * qty
        return round(subtotal * (1 - discount_percent / 100), 2)


class Order:
    def __init__(self, order_id, items, total):
        self.order_id, self.items, self.total = order_id, dict(items), total

    def to_dict(self):
        return {"order_id": self.order_id, "items": self.items, "total": self.total}


def place_order(cart, inventory, orders_path, discount_percent=0):
    if not cart.items:
        raise ValueError("Cannot place an empty order")
    for pid, qty in cart.items.items():
        if pid not in inventory.products:
            raise KeyError("Unknown product: " + pid)
        if inventory.products[pid].stock < qty:
            raise ValueError("Insufficient stock for " + pid)
    total = cart.total(inventory, discount_percent)
    order = Order(uuid.uuid4().hex, cart.items, total)
    path = Path(orders_path)
    old_stock = {pid: inventory.products[pid].stock for pid in cart.items}
    for pid, qty in cart.items.items():
        inventory.products[pid].stock -= qty
    try:
        existing = []
        if path.exists():
            existing = json.loads(path.read_text(encoding="utf-8"))
            if not isinstance(existing, list):
                raise ValueError("Orders file must contain a JSON list")
        existing.append(order.to_dict())
        path.parent.mkdir(parents=True, exist_ok=True)
        temp = path.with_suffix(path.suffix + "..tmp")
        temp.write_text(json.dumps(existing, indent=2), encoding="utf-8")
        temp.replace(path)
    except Exception:
        for pid, stock in old_stock.items():
            inventory.products[pid].stock = stock
        raise
    return order


class OrderSystemTests(unittest.TestCase):
    def setUp(self):
        self.inventory = Inventory([Product("P1", "Demo", 100, 5)])
        self.orders_path = "test_orders.json"

    def tearDown(self):
        p = Path(self.orders_path)
        if p.exists():
            p.unlink()

    def test_discount_and_stock_update(self):
        cart = Cart()
        cart.add("P1", 2)
        order = place_order(cart, self.inventory, self.orders_path, 10)
        self.assertEqual(order.total, 180)
        self.assertEqual(self.inventory.products["P1"].stock, 3)

    def test_out_of_stock_rejected_without_changes(self):
        cart = Cart()
        cart.add("P1", 6)
        with self.assertRaises(ValueError):
            place_order(cart, self.inventory, self.orders_path)
        self.assertEqual(self.inventory.products["P1"].stock, 5)

    def test_invalid_quantity_rejected(self):
        cart = Cart()
        with self.assertRaises(ValueError):
            cart.add("P1", 0)


if __name__ == "__main__":
    # Automatically create the required products.json file if it is missing
    products_file = Path("products.json")
    if not products_file.exists():
        sample_products = [
            {"product_id": "P1", "name": "Demo", "price": 100.0, "stock": 5}
        ]
        products_file.write_text(json.dumps(sample_products, indent=2), encoding="utf-8")

    inventory = Inventory.from_json("products.json")
    cart = Cart()
    cart.add("P1", 2)
    order = place_order(cart, inventory, "orders.json", 10)
    print("Total:", order.total)
    print("P1 stock:", inventory.products["P1"].stock)
    unittest.TextTestRunner(stream=sys.stdout, verbosity=2).run(
        unittest.defaultTestLoader.loadTestsFromTestCase(OrderSystemTests))
```
